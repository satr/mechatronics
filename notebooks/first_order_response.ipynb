# First-order response — interactive Python model

> **Educational simulation only.** This idealized model may be incorrect for a real machine; do not connect it to physical equipment. All values below are synthetic.

The **speed error** is $\varepsilon(t)=u-v(t)$. The natural exponential function is $\exp(z)=\mathrm{e}^{z}$, with Euler’s number $\mathrm{e}\approx2.71828$.


```python
import numpy as np
import matplotlib.pyplot as plt

# For local imports, install the optional dependencies from requirements.txt at the repository root.
```

## Step response

For constant target speed $u$, initial speed $v_0$, and time constant $\tau$:  
$\displaystyle v(t)=u+(v_0-u)\exp\left(-\frac{t}{\tau}\right)$.

Change `tau`, `target`, or `initial` and rerun the following cell. A smaller $\tau$ means a faster response **of this model**, not of a real motor.


```python
target = 1000.0  # rpm, synthetic target
initial = 0.0   # rpm
tau = 2.0       # s, model time constant

assert tau > 0
t = np.linspace(0.0, 4*tau, 401)
speed = target + (initial - target) * np.exp(-t/tau)

fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(t, speed, label="First-order response")
ax.axhline(target, linestyle="--", label="Target")
for n in (1, 2, 3):
    current_t = n*tau
    fraction = 1-np.exp(-n)
    current_speed = initial + (target-initial)*fraction
    ax.plot(current_t, current_speed, "o")
    ax.annotate(f"{100*fraction:.1f}%", (current_t, current_speed), xytext=(5,-15), textcoords="offset points")
ax.set(xlabel="Time (s)", ylabel="Speed (rpm)", title="Synthetic first-order speed response")
ax.grid(True)
ax.legend()
plt.show()
```

## Why 63.2%, 86.5%, and 95.0%?

The fraction completed is $f(t)=1-\exp(-t/\tau)$. Substituting $t=n\tau$ gives $f(n\tau)=1-\exp(-n)$.


```python
for n in (1, 2, 3):
    fraction = 1 - np.exp(-n)
    print(f"At {n} tau: {100*fraction:.4f}% of the change completed")
```
